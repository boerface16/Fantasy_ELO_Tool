# Talent ELO Sandbox

Replicates the 9D talent ELO pipeline locally for config experimentation.

**Workflow:**
1. Run Cell 1 once (imports + Supabase)
2. Run Cell 2 once (load data)
3. **Edit Cell 3 config dict**, then run Cell 3 + Cell 4 to recompute
4. Use Cells 5–7 to explore results

To tweak **event weights** (e.g. HR weight for Power), edit `config/multi_elo_config.yaml` directly, then rerun Cells 3–4.

```python
# Cell 1 — Imports & Environment
import sys, os, yaml, tempfile, importlib
sys.path.insert(0, '..')  # make src/ importable from notebooks/

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from dotenv import load_dotenv
from supabase import create_client

load_dotenv('../.env')
sb = create_client(os.environ['SUPABASE_URL'], os.environ['SUPABASE_KEY'])
print('Supabase connected')
```

```python
# Cell 2 — Load Data (run once)
# Pull 2026 PAs from opening day through 2026-04-08
# Note: leverage_index and is_risp are not stored in the DB.
# The engine derives clutch context from on_2b/on_3b/outs_when_up and defaults LI to 1.0.
PA_COLS = (
    'pa_id,game_date,batter_id,pitcher_id,result_type,'
    'on_2b,on_3b,outs_when_up,runner_id,season_year'
)
rows, offset = [], 0
while True:
    resp = (
        sb.table('plate_appearances')
        .select(PA_COLS)
        .eq('season_year', 2026)
        .lte('game_date', '2026-04-08')
        .order('game_date').order('pa_id')
        .range(offset, offset + 999)
        .execute()
    )
    if not resp.data:
        break
    rows.extend(resp.data)
    if len(resp.data) < 1000:
        break
    offset += 1000

pa_df = pd.DataFrame(rows)
pa_df['game_date'] = pd.to_datetime(pa_df['game_date']).dt.date
print(f'Loaded {len(pa_df):,} plate appearances across {pa_df["game_date"].nunique()} game days')
print(pa_df['result_type'].value_counts().head(12).to_string())

# Player lookup (name, team, position)
player_rows = sb.table('players').select('player_id,full_name,team,position').execute().data
players_df = pd.DataFrame(player_rows).set_index('player_id')
print(f'\n{len(players_df)} players in roster')
```

```python
# Cell 3 — Config (EDIT THIS TO TWEAK DIMENSION PARAMS)
# Change k_factor, scale, reliability_threshold, expected_divisor for any dimension.
# Then rerun this cell + Cell 4 to see the effect.
#
# To tweak event weights (e.g. HR=+1.0 for power), edit config/multi_elo_config.yaml
# then rerun Cell 4 (event weights are merged from that file automatically).

CONFIG = {
    'constants': {
        'default_elo': 1500.0,
        'elo_min': 500.0,
        'elo_max': 3000.0,
        'min_reliability': 0.3,
        'default_expected_divisor': 400.0,
    },
    'batter_dimensions': [
        # name, index, k_factor, scale, reliability_threshold, expected_divisor
        {'name': 'contact',    'index': 0, 'k_factor': 12.0,  'scale': 5.0,  'reliability_threshold': 100, 'expected_divisor': 127.0},
        {'name': 'power',      'index': 1, 'k_factor': 14.4,  'scale': 10.0, 'reliability_threshold': 100, 'expected_divisor': 218.0},
        {'name': 'discipline', 'index': 2, 'k_factor': 12.0,  'scale': 5.0,  'reliability_threshold': 100, 'expected_divisor': 143.0},
        {'name': 'speed',      'index': 3, 'k_factor': 36.0,  'scale': 4.0,  'reliability_threshold': 10,  'expected_divisor': 425.0},
        {'name': 'clutch',     'index': 4, 'k_factor': 18.0,  'scale': 6.0,  'reliability_threshold': 100, 'expected_divisor': 115.0},
    ],
    'pitcher_dimensions': [
        {'name': 'stuff',           'index': 0, 'k_factor': 12.0, 'scale': 5.0, 'reliability_threshold': 100, 'expected_divisor': 132.0},
        {'name': 'bip_suppression', 'index': 1, 'k_factor': 4.0,  'scale': 3.0, 'reliability_threshold': 100, 'expected_divisor': 181.0},
        {'name': 'command',         'index': 2, 'k_factor': 12.0, 'scale': 5.0, 'reliability_threshold': 100, 'expected_divisor': 148.0},
        {'name': 'clutch',          'index': 3, 'k_factor': 18.0, 'scale': 6.0, 'reliability_threshold': 100, 'expected_divisor': 123.0},
    ],
    # Composite weights — how dimensions combine into a single score.
    # Match production weights in src/fantasy/weekly_projection.py.
    # Clutch excluded (not in elo_lookup output used by projection).
    'composite_weights': {
        'batter':  {'contact': 0.30, 'power': 0.35, 'discipline': 0.25, 'speed': 0.10},
        'pitcher': {'stuff': 0.40, 'command': 0.35, 'bip_suppression': 0.25},
    },
}
print('Config ready — run Cell 4 to recompute')
```

```python
# Cell 4 — Run Engine (rerun after any config change)
import src.engine.multi_elo_config as _cfg_mod
import src.engine.multi_elo_engine as _eng_mod
import src.engine.talent_batch as _tb_mod
importlib.reload(_cfg_mod)
importlib.reload(_eng_mod)
importlib.reload(_tb_mod)

from src.engine.multi_elo_config import MultiEloConfig
from src.engine.talent_batch import TalentBatch

def build_config(user_cfg: dict, base_yaml: str = '../config/multi_elo_config.yaml') -> MultiEloConfig:
    """Merge user dimension params with event weights from the base YAML."""
    with open(base_yaml, encoding='utf-8') as f:
        base = yaml.safe_load(f)
    merged = base.copy()
    merged['constants'] = user_cfg['constants']
    merged['batter_dimensions'] = user_cfg['batter_dimensions']
    merged['pitcher_dimensions'] = user_cfg['pitcher_dimensions']
    tmp = tempfile.NamedTemporaryFile(mode='w', suffix='.yaml', delete=False, encoding='utf-8')
    yaml.dump(merged, tmp, allow_unicode=True)
    tmp.close()
    return MultiEloConfig(config_path=tmp.name)

config = build_config(CONFIG)

# All players start at ELO 1500 — no DB state loaded
batch = TalentBatch(config=config)
batch.process(pa_df)

# Current snapshot
records = batch.get_talent_player_records()
current_df = pd.DataFrame(records)

# OHLC history for line charts
ohlc_df = pd.DataFrame(batch.talent_daily_ohlc)
if not ohlc_df.empty:
    ohlc_df['game_date'] = pd.to_datetime(ohlc_df['game_date'])

print(f'Players with records: {current_df["player_id"].nunique()}')
print(f'OHLC rows: {len(ohlc_df)}')
print(f'Dimensions: {sorted(current_df["talent_type"].unique())}')

# Composite ELO — weighted sum of per-dimension season ELOs using CONFIG weights
# Falls back to production weights if Cell 3 hasn't been re-run in this kernel
_default_bw = {'contact': 0.30, 'power': 0.35, 'discipline': 0.25, 'speed': 0.10}
_default_pw = {'stuff': 0.40, 'command': 0.35, 'bip_suppression': 0.25}
bw = CONFIG.get('composite_weights', {}).get('batter', _default_bw)
pw = CONFIG.get('composite_weights', {}).get('pitcher', _default_pw)

def _weighted_sum(role_df, weights):
    return (
        role_df.assign(w=role_df['talent_type'].map(weights).fillna(0.0) * role_df['season_elo'])
        .groupby('player_id')['w'].sum()
        .reset_index()
        .rename(columns={'w': 'composite_elo'})
    )

batter_comp = _weighted_sum(current_df[current_df['player_role'] == 'batter'], bw)
batter_comp['player_role'] = 'batter'
pitcher_comp = _weighted_sum(current_df[current_df['player_role'] == 'pitcher'], pw)
pitcher_comp['player_role'] = 'pitcher'
composite_df = pd.concat([batter_comp, pitcher_comp], ignore_index=True)

print(f'Composite ELO computed: {len(composite_df)} players')
_preview = composite_df.merge(players_df[['full_name']].reset_index(), on='player_id', how='left')
_preview['full_name'] = _preview.apply(
    lambda r: r['full_name'] if pd.notna(r['full_name']) else f"ID:{int(r['player_id'])}", axis=1
)
print(_preview.sort_values('composite_elo', ascending=False).head(5)
      [['full_name', 'player_role', 'composite_elo']].to_string(index=False))
```

```python
# Cell 5 — Leaderboards
# Change role ('batter'/'pitcher') and dimension, then re-run this cell.

def show_leaderboard(role: str, dimension: str, n: int = 25):
    subset = current_df[
        (current_df['player_role'] == role) &
        (current_df['talent_type'] == dimension)
    ].copy()
    subset = subset.merge(
        players_df[['full_name', 'team']].reset_index(),
        on='player_id', how='left'
    )
    top = subset.sort_values('season_elo', ascending=False).head(n).copy()
    top.index = range(1, len(top) + 1)
    styled = (
        top[['full_name', 'team', 'season_elo', 'event_count']]
        .rename(columns={'full_name': 'Player', 'team': 'Team', 'season_elo': 'ELO', 'event_count': 'Events'})
        .style
        .background_gradient(subset='ELO', cmap='RdYlGn', vmin=1400, vmax=1700)
        .format({'ELO': '{:.1f}'})
        .set_caption(f'Top {n} {role.title()}s — {dimension.title()}')
    )
    display(styled)

# --- Edit these calls ---
show_leaderboard('batter', 'power')
show_leaderboard('batter', 'speed')
show_leaderboard('batter', 'contact')
show_leaderboard('pitcher', 'stuff')
show_leaderboard('pitcher', 'command')

def show_composite_leaderboard(role: str, n: int = 25):
    subset = composite_df[composite_df['player_role'] == role].copy()
    subset = subset.merge(
        players_df[['full_name', 'team']].reset_index(),
        on='player_id', how='left'
    )
    # Also join PA count from current_df
    pa_counts = (
        current_df[current_df['player_role'] == role]
        .groupby('player_id')['pa_count'].max()
        .reset_index()
    )
    subset = subset.merge(pa_counts, on='player_id', how='left')
    top = subset.sort_values('composite_elo', ascending=False).head(n).copy()
    top.index = range(1, len(top) + 1)
    styled = (
        top[['full_name', 'team', 'composite_elo', 'pa_count']]
        .rename(columns={'full_name': 'Player', 'team': 'Team', 'composite_elo': 'Composite ELO', 'pa_count': 'PA/BFP'})
        .style
        .background_gradient(subset='Composite ELO', cmap='RdYlGn', vmin=1450, vmax=1650)
        .format({'Composite ELO': '{:.1f}'})
        .set_caption(f'Top {n} {role.title()}s — Composite ELO')
    )
    display(styled)

show_composite_leaderboard('batter')
show_composite_leaderboard('pitcher')
```

```python
# Cell 5b — Player Name → ID Search
# Use this to find the numeric player_id needed for plot_elo_history() and plot_radar().
# Copy the player_id number from the output and paste it into the function calls below.

def find_player(name_fragment: str, position: str = None):
    """Search players_df by name. Optionally filter by position (e.g. 'SP', 'CF', '1B')."""
    mask = players_df['full_name'].str.contains(name_fragment, case=False, na=False)
    if position:
        mask &= players_df['position'].str.upper().eq(position.upper())
    result = players_df[mask][['full_name', 'team', 'position']].reset_index()
    result.columns = ['player_id', 'full_name', 'team', 'position']
    return result

# --- Edit name fragments below to search ---
display(find_player('Judge'))
display(find_player('Skenes'))
display(find_player('Henderson'))
display(find_player('Ohtani'))
display(find_player('Fried'))
```

```python
# Cell 5c -- Find players with ELO records but missing from players_df (no name)
# Cause: player appeared in PA data but wasn't registered in Supabase players table

power_df = current_df[
    (current_df['player_role'] == 'batter') &
    (current_df['talent_type'] == 'power')
].merge(
    players_df[['full_name', 'team', 'position']].reset_index(),
    on='player_id', how='left'
)

missing_names = power_df[power_df['full_name'].isna()].sort_values('season_elo', ascending=False)
print(f"{len(missing_names)} batters with power ELO but no name in players_df:")
display(missing_names[['player_id', 'season_elo', 'event_count', 'pa_count']])

# Look up names from MLB Stats API
import requests

print()
for pid in missing_names['player_id'].tolist():
    r = requests.get(f"https://statsapi.mlb.com/api/v1/people/{pid}").json()
    person = r['people'][0]
    print(f"{pid}: {person['fullName']}  ({person.get('primaryPosition', {}).get('abbreviation', '?')})",)
```

```python
# Cell 6 — ELO History Line Charts
# Find player IDs via: players_df[players_df['full_name'].str.contains('Judge')]

def plot_elo_history(player_ids: list, dimension: str):
    """Line chart of close ELO over time for selected players."""
    fig, ax = plt.subplots(figsize=(13, 5))
    for pid in player_ids:
        name = players_df.loc[pid, 'full_name'] if pid in players_df.index else str(pid)
        pdata = ohlc_df[(ohlc_df['player_id'] == pid) & (ohlc_df['talent_type'] == dimension)]
        if pdata.empty:
            print(f'  No data for {name} / {dimension}')
            continue
        ax.plot(pdata['game_date'], pdata['close'], label=name, linewidth=2)
    ax.axhline(1500, color='gray', linestyle='--', alpha=0.5, linewidth=1, label='Baseline 1500')
    ax.set_title(f'{dimension.title()} ELO History — 2026', fontsize=14)
    ax.set_ylabel('ELO')
    ax.legend()
    ax.grid(alpha=0.25)
    ax.yaxis.set_major_formatter(mticker.FormatStrFormatter('%d'))
    fig.tight_layout()
    plt.show()

# Helper: find a player ID
def find_player(name_fragment: str):
    return players_df[players_df['full_name'].str.contains(name_fragment, case=False)][['full_name','team','position']]

# --- Example usage ---
#find_player('Judge')
plot_elo_history([592450, 683002], 'power')   # Aaron Judge, Gunnar Henderson
plot_elo_history([608331, 694973], 'stuff')   # Paul Skenes, Zack Wheeler
```

```python
# Cell 7 — Radar Chart + Distribution

def plot_radar(player_id: int, role: str = 'batter'):
    """Spider/radar chart of all talent dimensions for a player."""
    dims = (
        ['contact', 'power', 'discipline', 'speed', 'clutch']
        if role == 'batter'
        else ['stuff', 'bip_suppression', 'command', 'clutch']
    )
    elos = []
    for d in dims:
        row = current_df[
            (current_df['player_id'] == player_id) &
            (current_df['player_role'] == role) &
            (current_df['talent_type'] == d)
        ]
        elos.append(float(row['season_elo'].iloc[0]) if not row.empty else 1500.0)

    N = len(dims)
    angles = [n / N * 2 * np.pi for n in range(N)]
    angles_plot = angles + angles[:1]
    elos_plot = elos + elos[:1]

    fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
    ax.plot(angles_plot, elos_plot, linewidth=2, color='steelblue')
    ax.fill(angles_plot, elos_plot, alpha=0.25, color='steelblue')
    ax.set_xticks(angles)
    ax.set_xticklabels([d.replace('_', '\n') for d in dims], fontsize=10)
    ax.set_ylim(1300, 1750)
    ax.yaxis.set_major_formatter(mticker.FormatStrFormatter('%d'))
    name = players_df.loc[player_id, 'full_name'] if player_id in players_df.index else str(player_id)
    ax.set_title(f'{name}\n{role.title()} Talent Radar — 2026', pad=20, fontsize=12)
    fig.tight_layout()
    plt.show()


def plot_distribution(role: str, dimension: str, min_events: int = 10):
    """Histogram of ELO distribution for a dimension (qualified players only)."""
    data = current_df[
        (current_df['player_role'] == role) &
        (current_df['talent_type'] == dimension) &
        (current_df['event_count'] >= min_events)
    ]['season_elo']
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.hist(data, bins=40, color='steelblue', edgecolor='white', alpha=0.85)
    ax.axvline(1500, color='red', linestyle='--', linewidth=1.5, label='Baseline 1500')
    ax.axvline(data.mean(), color='gold', linestyle='-', linewidth=1.5, label=f'Mean {data.mean():.0f}')
    ax.set_title(f'{role.title()} {dimension.title()} ELO Distribution (min {min_events} events, n={len(data)})')
    ax.set_xlabel('ELO')
    ax.set_ylabel('Players')
    ax.legend()
    ax.grid(alpha=0.25)
    fig.tight_layout()
    plt.show()


# --- Example usage ---
plot_radar(592450, role='batter')      # Aaron Judge
plot_radar(673357, role='pitcher')     # Paul Skenes
plot_distribution('batter', 'power')
plot_distribution('batter', 'contact')
plot_distribution('batter', 'discipline')
plot_distribution('batter', 'clutch')
plot_distribution('pitcher', 'stuff')
plot_distribution('pitcher', 'command')
plot_distribution('pitcher', 'clutch')
plot_distribution('pitcher', 'bipsupression')
```
